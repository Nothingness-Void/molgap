```python
import json, os, pathlib, platform, sys, time
import jax

devices = [str(device) for device in jax.devices()]
report = {
    'format': 'molgap-v5-kaggle-tpu-notebook-allocation-v1',
    'created_at_unix': time.time(),
    'python': sys.version,
    'platform': platform.platform(),
    'jax': jax.__version__,
    'jax_backend': jax.default_backend(),
    'devices': devices,
    'tpu_visible': any('TPU' in device.upper() for device in devices),
    'selected_environment': {
        key: os.environ.get(key)
        for key in ('KAGGLE_KERNEL_RUN_TYPE', 'TPU_NAME', 'COLAB_TPU_ADDR')
        if os.environ.get(key)
    },
}
pathlib.Path('/kaggle/working/tpu_allocation_report.json').write_text(
    json.dumps(report, indent=2, sort_keys=True), encoding='utf-8'
)
print(json.dumps(report, indent=2, sort_keys=True), flush=True)
if not report['tpu_visible']:
    raise RuntimeError(f'TPU was requested but JAX devices are {devices}')
```
